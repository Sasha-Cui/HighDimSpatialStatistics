```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the loss history data
l=0
i=1
j=0
for j in range(3):
    file_path = os.path.expanduser(f"~/project/43_estimation_results/{l}/loss_histories_dataset_{i}_feature_{j}.csv")
    loss_history = pd.read_csv(file_path)
    
    # Plot the loss history
    plt.figure(figsize=(10, 6))
    plt.plot(loss_history, label="Loss", marker="o")
    plt.title(f"Loss History for Hyperparam {l}, Dataset {i}, Feature {j}")
    plt.xlabel("Iteration")
    plt.ylabel("Loss Value")
    plt.xscale("log")
    plt.legend()
    plt.grid()
    plt.show()
```
